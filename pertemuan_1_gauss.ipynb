# Penyelesaian SPL dengan Metode Eliminasi Gauss

**Bagian Fraction (sel 2):**
- `Fraction`: tipe data pecahan dari modul `fractions`, supaya hasil perhitungan eksak (bukan desimal pembulatan).
- `tampil_matriks(M)`: mencetak matriks dengan elemen berbentuk pecahan.

**Fungsi-fungsi Gauss (sel 3):**
- `buat_augmented(A, b)`: menggabungkan matriks koefisien A dan vektor konstanta b menjadi matriks augmented [A | b].
- `eliminasi_maju(M)`: mengubah matriks augmented menjadi bentuk segitiga atas.
- `substitusi_mundur(M)`: menghitung nilai variabel dari baris terakhir ke baris pertama.
- `gauss(A, b)`: fungsi utama yang memanggil ketiga fungsi di atas.

**Eksekusi (sel 4):** mendefinisikan SPL (A dan b), menjalankan `gauss`, lalu menampilkan solusi.

```python
from fractions import Fraction

def tampil_matriks(M):
    """Cetak matriks dengan elemen berbentuk pecahan."""
    for baris in M:
        print("  ".join(f"{str(e):>6}" for e in baris))
    print()
```

“dari pustaka fractions, ambil Fraction.” Dengan Fraction, 1/3 tersimpan sebagai pecahan eksak, bukan 0.3333333.

`for e in baris`: ambil setiap elemen e dalam baris itu.
`str(e)`: ubah elemen menjadi teks, misalnya Fraction(1,2) menjadi "1/2".
`f"{...:>6}"`: sisipkan teks itu dengan rata kanan selebar 6 karakter.
`" ".join(...)`: gabungkan semua elemen dengan dua spasi di antaranya.
`print(...)`: cetak hasilnya sebagai satu baris.

```python
def buat_augmented(A, b):
    """Gabungkan A dan b menjadi matriks augmented berisi Fraction."""
    n = len(b)
    return [[Fraction(A[i][j]) for j in range(n)] + [Fraction(b[i])]
            for i in range(n)]
```

`n` adalah jumlah persamaan, dihitung dari panjang b. Pada contoh, `n = 3`.
Perulangan luar `for i in range(n)` membuat satu baris untuk tiap `i (0, 1, 2)`.
Bagian dalam `[Fraction(A[i][j]) for j in range(n)]` mengambil semua koefisien baris ke-i dan mengubahnya menjadi Fraction.
`[Fraction(b[i])]` menempelkan konstanta `b[i]` di ujung kanan.

```python
def eliminasi_maju(M):
    n = len(M)
    for k in range(n):
        # Pivoting: cari baris dengan elemen tidak nol di kolom k
        p = next((i for i in range(k, n) if M[i][k] != 0), None)
        if p is None:
            raise ValueError("SPL tidak punya solusi tunggal")
        M[k], M[p] = M[p], M[k]

        # Nolkan elemen di bawah pivot
        for i in range(k + 1, n):
            f = M[i][k] / M[k][k]
            for j in range(k, n + 1):
                M[i][j] -= f * M[k][j]
    return M
```

`k` adalah nomor langkah sekaligus indeks baris dan kolom pivot. Pada `k = 0` pivotnya `M[0][0]`, pada `k = 1` pivotnya `M[1][1]`, dan seterusnya. dari baris `k` sampai bawah, cari `i` pertama yang `M[i][k]`-nya bukan nol. Kalau tidak ada, hasilnya `None`.” `next(..., None)` mengambil temuan pertama.

```python
def substitusi_mundur(M):
    n = len(M)
    x = [Fraction(0)] * n
    for i in range(n - 1, -1, -1):
        s = sum(M[i][j] * x[j] for j in range(i + 1, n))
        x[i] = (M[i][n] - s) / M[i][i]
    return x
```

```python
def gauss(A, b):
    M = buat_augmented(A, b)
    print("Matriks augmented awal:")
    tampil_matriks(M)

    M = eliminasi_maju(M)
    print("Setelah eliminasi maju (segitiga atas):")
    tampil_matriks(M)

    return substitusi_mundur(M)
```

```python
# SPL:  2x +  y -  z =   8
#      -3x -  y + 2z = -11
#      -2x +  y + 2z =  -3
A = [[2, 1, -1],
     [-3, -1, 2],
     [-2, 1, 2]]
b = [8, -11, -3]

hasil = gauss(A, b)

print("Solusi:")
for i, nilai in enumerate(hasil, start=1):
    print(f"x{i} = {nilai}")
```
